Numpy

```python
import numpy as np
students=np.array([
    [85,90,78],
    [60,49,70],
    [95,92,98],
    [40,58,45],
    [72,68,75]
])
students
```

```python
avg_marks=np.mean(students,axis=1)
print("Average marks of each student:",avg_marks)
```

```python
avg_marks=np.mean(students,axis=0)
print("Average marks of each student:",avg_marks)
```

```python
topper=np.argmax(avg_marks)
print("Topper is:",topper)
```

```python
failed_students=np.where(avg_marks<50)
print("Failed students are:",failed_students[0]+1)
```

```python
for i, avg in enumerate(avg_marks):
  status="pass" if avg>50 else "fail"
  print(f"Student {i+1}: Avg = {avg:.2f}|{status}")
```

```python
#Bar chart
months=["Jan","Feb","Mar","April"]
sales=[250,300,400,350]
plt.bar(months,sales,color='pink')
plt.title('Monthly sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
plt.barh(months,sales,color='violet')
plt.title('Monthly sales')
plt.xlabel('Sales')
plt.ylabel('Months')
plt.grid(True)
plt.show()
```

```python
#Scatter Plot(comparing height vs weight)
height = [165, 170, 160, 180, 175]
weight = [65, 70, 55, 80, 75]
plt.scatter(height, weight, color='lime')
plt.title('Height vs Weight')
plt.xlabel('Height(cm)')
plt.ylabel('Weight(kg)')
plt.grid(True)
plt.show()
```

```python
#Pie chart(markrt share)
brands=['Brand A','Brand B','Brand C','Brand D']
share=[30,20,40,10]
plt.pie(share,labels=brands,autopct='%1.1f%%',startangle=50,colors=['pink','violet','skyblue','yellow'])
plt.title('Market Shares')
plt.axis('equal')
plt.show()
```

```python
#Multiple plot in one figure
x=[1,2,3,4]
y=[1,4,9,16]
y2=[1,8,27,64]
plt.plot(x,y2,label='Square',color="lime")
plt.plot(x,y,label='Linear',color="purple")
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.title('Multiple Plots')
plt.grid(True)
plt.show()
```

```python
#Subplots
fig,axs=plt.subplots(2,2)

axs[0,0].plot(x,y2)
axs[0,0].set_title('Square')

axs[0,1].bar(months,sales)
axs[0,1].set_title('Sales')

axs[1,0].scatter(height,weight)
axs[1,0].set_title('Scatter')

axs[1,1].pie(share,labels=brands,autopct='%1.1f%%',startangle=50,colors=['pink','violet','skyblue','yellow'])
axs[1,1].set_title('Pie')

plt.tight_layout()
plt.show()
```
